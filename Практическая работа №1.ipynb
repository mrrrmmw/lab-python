# Задание 3. Расчёт сложного процента

Даны переменные `summa, percent, years`. Вычислить итоговую сумму по формуле:

$$S= summa\cdot (1 + percent/100)^{years}$$

Вывести результат с двумя знаками после запятой.

# Код программы:

````python
summa = 10000
percent = 5
years = 3

S = summa * (1 + percent / 100) ** years

print(f"Итоговая сумма: {S:.2f}")
````

5. Арифметические операции в Python

Оператор `/` выполняет обычное деление, а оператор `//` — деление с округлением результата вниз до ближайшего целого значения.

8. Что означает, что выражение возвращает значение?

Выражение — это комбинация значений, переменных и операторов, при вычислении которой получается определённый результат.

9. Операторы сравнения и логические операторы

Операторы сравнения позволяют сопоставлять значения:

* `==`	Равно
* `!=`	Не равно
* `<`	Меньше
* `>`	Больше
* `<=`	Меньше или равно
* `>=`	Больше или равно

Логические операторы:
- and — истина, если оба условия истинны;
- or — истина, если хотя бы одно условие истинно;
- not — инвертирует логическое значение.

12. Порядок выполнения операций в Python

Порядок выполнения операций определяет, какие действия выполняются раньше других. Например, умножение выполняется раньше сложения, а скобки позволяют задать собственный порядок вычислений.

13. Множественное присваивание в Python
Множественное присваивание позволяет присваивать значения нескольким переменным в одной строке.

Присваивание нескольких значений:
```
a, b, c = 10, 20, 30
print(a, b, c)
```

Обмен значений двух переменных:
```
a = 5
b = 10

a, b = b, a
print(a, b)
```

Распаковка списка:
```
numbers = [1, 2, 3]
a, b, c = numbers

print(a, b, c)
```

14. Изменяемые и неизменяемые типы данных

Неизменяемые типы — это типы, объекты которых нельзя изменить после создания. К ним относятся `int`, `float`, `bool` и `str`.
Изменяемые типы позволяют менять содержимое объекта после его создания. Например, к ним относятся `list` и `dict`.
