## Задание 1
Пользователь вводит число. Программа определяет, положительное оно, отрицательное или равно нулю, и выводит результат. Используйте условный оператор.

Решение

## Задание 2
Пользователь вводит число n. Программа определяет, является ли оно совершенным (равно сумме всех своих делителей, кроме самого себя). Используйте цикл for и условный оператор. Пример: 6 = 1 + 2 + 3.

````python
# Ввод числа
n = int(input('Введите число: '))

# Сумма делителей
s = 0

# Перебор делителей числа
for i in range(1, n):
    if n % i == 0:
        s += i

# Проверка числа
if n > 0 and s == n:
    print('Число совершенное')
else:
    print('Число не является совершенным')
````

## Задание 3
Пользователь вводит число n. Программа вычисляет золотое сечение двумя способами: через math.sqrt и через предел последовательности Фибоначчи $F_{n+1}/F_n$. Выведите оба результата с 10 знаками и разность между ними.


````python
import math

# Ввод числа
n = int(input('Введите количество чисел Фибоначчи: '))

# Вычисление золотого сечения через math.sqrt
phi_math = (1 + math.sqrt(5)) / 2

# Вычисление чисел Фибоначчи
a = 0
b = 1

for i in range(n):
    a, b = b, a + b

# Вычисление золотого сечения через последовательность Фибоначчи
phi_fib = b / a

# Вычисление разности
difference = abs(phi_math - phi_fib)

# Вывод результатов
print(f'Золотое сечение через math.sqrt: {phi_math:.10f}')
print(f'Золотое сечение через Фибоначчи: {phi_fib:.10f}')
print(f'Разность: {difference:.10f}')
````

1. В чём принципиальное отличие циклов `for` и `while` в Python? Приведите примеры задач.

Цикл for используется, когда нужно перебрать элементы последовательности или выполнить определённое количество повторений.

Пример: вывести числа от 1 до 5.
```
for i in range(1, 6):    print(i)
```
Цикл `while` используется, когда количество повторений заранее неизвестно и выполнение зависит от условия.

Пример: запрашивать пароль, пока пользователь не введёт правильный.
```
password = ''while password != '1234':    password = input('Введите пароль: ')print('Доступ разрешён')
```


2. Как работает блок `else` внутри цикла `(for/while)`?

Блок `else` выполняется, если цикл завершился обычным образом, без использования оператора break.

Пример, когда else срабатывает:
```
for i in range(3):    print(i)else:    print('Цикл завершён')
```


Пример, когда else не срабатывает:
```
for i in range(5):    if i == 2:        break    print(i)else:    print('Цикл завершён')
```


Во втором примере цикл прерывается оператором break, поэтому блок else не выполняется.

3. Объясните разницу между операторами `break` и `continue`.

break полностью прекращает выполнение цикла, а continue пропускает оставшуюся часть текущей итерации и переходит к следующей.

Пример:
```
i = 0while i < 6:    i += 1    if i == 2:        continue    if i == 5:        break    print(i)
```


Результат:
1
3
4


Объяснение:
- При i = 1 число выводится.
- При i = 2 срабатывает continue, поэтому число пропускается.
- При i = 3 и i = 4 числа выводятся.
- При i = 5 срабатывает break, и цикл завершается.
4. Какие типичные ошибки возникают при организации циклов в Python?

Основные ошибки:

1. Бесконечный цикл — условие никогда не становится ложным.

Ошибочный код:
`i = 0while i < 5:    print(i)`



Исправленный код:
`i = 0while i < 5:    print(i)    i += 1`



2. Неверный диапазон range() — верхняя граница не включается.

Ошибочный код:
`for i in range(1, 5):    print(i)`



Если нужно вывести числа от 1 до 5, исправленный код:
`for i in range(1, 6):    print(i)`



3. Неправильное условие выхода — цикл завершается раньше или позже необходимого.

Ошибочный код:
`i = 1while i != 5:    print(i)    i += 2`



Исправленный код:
`i = 1while i < 5:    print(i)    i += 1`



5. Как устроена условная конструкция if/elif/else?

Конструкция `if/elif/else` позволяет проверять несколько условий. Сначала проверяется if, затем при необходимости — elif. Если ни одно условие не выполнилось, выполняется else.

Пример:
```
x = int(input('Введите число: '))if x < 0:    print('Меньше нуля')elif x <= 10:    print('От 0 до 10')else:    print('Больше 10')
```


Порядок условий важен, потому что Python проверяет их сверху вниз и выполняет только первый подходящий блок.

6. Что такое логические операторы and, or, not?
- and — возвращает истину, если оба условия истинны.
- or — возвращает истину, если хотя бы одно условие истинно.
- not — меняет значение условия на противоположное.
Условие, которое истинно, когда число чётное и находится от 10 до 20 либо нечётное и больше 100:
```
(x % 2 == 0 and 10 <= x <= 20) or (x % 2 != 0 and x > 100)
```

9. Как работают встроенные функции min, max, sum, abs, round?
- min() — находит минимальное значение.
- max() — находит максимальное значение.
- sum() — вычисляет сумму элементов.
- abs() — возвращает абсолютное значение числа.
- round() — округляет число до указанного количества знаков после запятой.

Примеры:
```
numbers = [4, 8, 2, 10]print(min(numbers))  # 2print(max(numbers))  # 10print(sum(numbers))  # 24print(abs(-15))      # 15print(round(3.14159, 2))  # 3.14
```

Функции min(), max() и sum() работают с последовательностями чисел, abs() — с числовыми значениями, а round() — с целыми и вещественными числами. Результат зависит от функции: это минимальное или максимальное значение, сумма либо числовой результат.
