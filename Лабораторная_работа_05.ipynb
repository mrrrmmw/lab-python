## Задание

В файле `pr5_1.txt` содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
**0 до 10 000** включительно. Определите количество пар, в которых
оба элемента меньше, чем среднее арифметическое всех чисел в
файле, и хотя бы **один из двух элементов оканчивается на 19**. В
ответе запишите два числа: сначала количество найденных пар, а
затем – максимальную сумму элементов таких пар. В данной задаче
под парой подразумевается два идущих подряд элемента
последовательности.

## Решение

Чтение данных в список
Создадим пустой список `data` и добавим в него все числа из файла:

````python
data = []

with open('pr5_1.txt') as F:
    for line in F:
        digit = int(line)
        data.append(digit)

print(len(data))
print(data[:10])
````

Здесь каждая строка файла преобразуется из строкового типа в целое число с помощью `int()` и добавляется в список `data`

````python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)

print('Среднее арифметическое =', sr)
````

````python
# Подсчет количества подходящих пар
counter = 0
max_sum = 0

for i in range(len(data) - 1):
    x, y = data[i], data[i + 1]

    if x < sr and y < sr and (x % 100 == 19 or y % 100 == 19):
        counter += 1
        max_sum = max(max_sum, x + y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
````

Выполните задание в соответствии с индивидуальным вариантом

````python
data = []

with open('pr5_1.txt') as F:
    for line in F:
        digit = int(line)
        data.append(digit)

print('Количество чисел =', len(data))
print('Первые 10 чисел:', data[:10])

sr = sum(data) / len(data)

print('Среднее арифметическое =', sr)

counter = 0
max_sum = 0

for i in range(len(data) - 1):
    x, y = data[i], data[i + 1]

    if x < sr and y < sr and (x % 100 == 19 or y % 100 == 19):
        counter += 1
        max_sum = max(max_sum, x + y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
````
