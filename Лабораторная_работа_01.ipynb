## Задание

Необходимо вычислить значение выражения:
$$Z=\frac{35}{f}+y\cdot{f}-\frac{f+y}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $f, y$. Переменная, которая должна быть вычислена на основе независимых: $Z$.

````python
f=5
y=30.67
print("f = ", f, ", type(f) = ", type(f))
print("y = ", y, ", type(y) = ", type(y))
````

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

````python
Z =35/f+y*f-(f+y)/4
print('Z=',Z)
print('Type of Z: ', type(Z))
````

17.   ЦП понимает инструкции, которые написаны только на одном языке. Как называется этот язык?

Этот язык называется машинным языком или машинным кодом.
Инструкции машинного языка представлены в виде кодов, которые непосредственно понимает центральный процессор.

18.   Как называется тип памяти, в которую программа должна копироваться при каждом её исполнении центральным процессором?

Это оперативная память (ОЗУ, RAM).
Перед выполнением программа загружается из постоянного хранилища в оперативную память.

19.   Как называется процесс, в котором участвует ЦП, когда он исполняет инструкции в программе?

Этот процесс называется выполнением программы или исполнением инструкций.
ЦП последовательно получает инструкции, декодирует их и выполняет соответствующие операции.

20.   Что такое язык ассемблера?
Язык ассемблера — это низкоуровневый язык программирования, в котором машинные инструкции представлены специальными символьными обозначениями — мнемониками.
Например, вместо непосредственного машинного кода используются команды вроде MOV, ADD, SUB.
