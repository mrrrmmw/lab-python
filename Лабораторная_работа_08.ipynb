## Задание

Вася составляет 6-буквенные слова из букв слова ДЕРЕВО так, что буквы на
позициях одинаковой чётности расположены в алфавитном порядке. Сколько
различных слов может составить Вася?

## Решение

Для решения задачи воспользуемся библиотекой `itertools`, которая позволяет перебирать различные перестановки букв. Функция `permutations()` создаёт все возможные перестановки, а `set()` позволяет убрать повторяющиеся варианты, так как в слове «ДЕРЕВО» две буквы «Е».

Сначала получим все перестановки букв, затем проверим условие задачи: буквы на нечётных позициях `(1, 3, 5)` и чётных позициях `(2, 4, 6)` должны располагаться в алфавитном порядке.

```python
import itertools

# Исходное слово
alf = 'ДЕРЕВО'

# Счётчик подходящих слов
number = 0

# Перебираем все различные перестановки
for word in sorted(set(itertools.permutations(alf))):
    word = ''.join(word)

    # Проверяем порядок букв на нечётных и чётных позициях
    if (word[0] <= word[2] <= word[4] and
        word[1] <= word[3] <= word[5]):
        number += 1
        # print(number, word)

print(number)
```

Для решения задачи использовалась библиотека itertools и функция permutations(), которая позволяет получить все возможные перестановки букв заданного слова. С помощью set() были удалены повторяющиеся перестановки, возникающие из-за одинаковых букв. Затем с помощью условного оператора if были отобраны слова, в которых буквы на нечётных и чётных позициях расположены в алфавитном порядке.

Модуль в Python — это файл с программным кодом, в котором содержатся функции, классы и переменные. Модули позволяют разделять программу на отдельные части и повторно использовать готовый код.
