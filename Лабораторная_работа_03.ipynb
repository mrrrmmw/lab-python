## Задание

Напишите функцию, которая по *трём переданным натуральным* числам возвращает, могут ли эти числа являться членами *арифметической прогрессии*.

Если «да», то функция возвращает **True**, в противном случае — **False**.


## Решение

### Объявление функции

````python
def is_arithmetic(a, b, c):
    numbers = sorted([a, b, c])
    return numbers[1] - numbers[0] == numbers[2] - numbers[1]
````

Ответы на контрольные вопросы


1.   Чем отличается функция с возвратом значения от функции без возврата значения?

Функция с возвратом значения после выполнения возвращает результат с помощью оператора return. Этот результат можно сохранить в переменную или использовать в другом выражении.

Функция без возврата значения выполняет определённые действия, но не возвращает результат для дальнейшего использования.

2.   Что такое библиотечная функция?

Библиотечная функция — это готовая функция, которая предоставляется языком программирования или подключаемой библиотекой и которую можно использовать в своей программе.

3. Почему библиотечные функции похожи на «чёрные ящики»?

Потому что пользователю не обязательно знать внутреннее устройство функции. Достаточно знать её название, параметры и результат работы.

4. Что делает приведённая ниже инструкция?


```
x = random.randint(1, 100)
```
Функция `randint()` из модуля random генерирует случайное целое число от 1 до 100 включительно.

6. Для чего используется следующий вызов библиотечной функции?


```

```
Функция `seed()` устанавливает начальное значение генератора псевдослучайных чисел.

Это позволяет при одинаковом начальном значении получать одинаковую последовательность псевдослучайных чисел, что удобно, например, при тестировании программ.


7. Каким образом функции помогают повторно использовать код в программе?

Функции позволяют объединить часто используемый набор команд в отдельный блок, а затем многократно вызывать его в разных местах программы.

8. Назовите и опишите две части определения функции.

Определение функции состоит из двух основных частей:

* Заголовок функции — начинается с ключевого слова def, содержит имя функции и её параметры.
* Тело функции — блок команд с отступом, который выполняется при вызове функции.

9. Как у функции задать параметры по умолчанию?

Параметру можно присвоить значение непосредственно в заголовке функции.

Например:


```
def hello(name="Пользователь"):
    print("Привет,", name)
```
Если вызвать:


```
hello()
```
будет использовано значение "Пользователь".
Если передать собственное значение:

```
hello("Анна")
```
будет использовано "Анна".

Рекурсия — это способ организации функции, при котором функция вызывает саму себя.
