# Лабораторная работа 3. Пользовательские функции

## Задание

Напишите функцию, которая по *трём переданным натуральным* числам возвращает, могут ли эти числа являться членами *арифметической прогрессии*.

Если «да», то функция возвращает **True**, в противном случае — **False**.


## Решение

### Объявление функции

```python
def is_arithmetic(a, b, c):
    numbers = sorted([a, b, c])
    return numbers[1] - numbers[0] == numbers[2] - numbers[1]
```

### Демонстрация использования функции

```python
def is_arithmetic(a, b, c):
    numbers = sorted([a, b, c])
    return numbers[1] - numbers[0] == numbers[2] - numbers[1]
x, y, z = 2, 4, 6
print(f'Числа {x}, {y}, {z} являются членами арифметической прогрессии: {is_arithmetic(x, y, z)}')

x, y, z = 3, 7, 10
print(f'Числа {x}, {y}, {z} являются членами арифметической прогрессии: {is_arithmetic(x, y, z)}')

x, y, z = 15, 10, 5
print(f'Числа {x}, {y}, {z} являются членами арифметической прогрессии: {is_arithmetic(x, y, z)}')
```
