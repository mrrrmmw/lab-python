## Решение

Реализуем чтение строки из файла

````python
mystr = open('pr6_1.txt').read().strip()

print(len(mystr))
print(mystr[:50])
````

Теперь ответим на поставленные вопросы, используя строку `mystr`

````python
counter = 0

for i in range(2, len(mystr)):
    if mystr[i] == 'D' and mystr[i-1].isdigit() and mystr[i-2].isdigit():
        counter += 1

print(f'Количество букв D: {counter}')
````
