## Задание 1
Напишите функцию max2(a, b), возвращающую большее из двух чисел без использования встроенной max. Протестируйте на числах и на строках.

````python
# Функция для поиска большего значения
def max2(a, b):
    if a > b:
        return a
    else:
        return b

# Проверка на числах
print(max2(10, 5))
print(max2(3, 8))

# Проверка на строках
print(max2('яблоко', 'банан'))
print(max2('кот', 'собака'))
````

Функция сравнивает два значения с помощью оператора >. Если первое значение больше второго, возвращается первое, иначе — второе. Строки сравниваются в лексикографическом порядке, то есть по кодам символов.

## Задание 2
Напишите функцию create_user(name, role="user", active=True) и сделайте несколько вызовов: с пропусками параметров, с именованными в произвольном порядке, с явным переопределением значений по умолчанию.

````python
# Функция создания пользователя
def create_user(name, role="user", active=True):
    return {
        "name": name,
        "role": role,
        "active": active
    }

# Вызов со значениями по умолчанию
print(create_user("Алексей"))

# Вызов с указанием роли
print(create_user("Мария", "admin"))

# Вызов с именованными параметрами в произвольном порядке
print(create_user("Иван", active=False, role="moderator"))

# Вызов с переопределением значений по умолчанию
print(create_user("Ольга", role="admin", active=False))
````

Функция create_user() принимает три параметра. Для параметров role и active заданы значения по умолчанию. Если при вызове функции их не указывать, Python использует значения user и True. Именованные аргументы можно передавать в произвольном порядке.

## Задание 3
Напишите рекурсивную функцию перевода целого числа в двоичную запись (в виде строки). Почему базовый случай — не 1, а 0 (или деление до нуля)?

````python
# Функция перевода числа в двоичную систему
def to_binary(n):
    if n < 2:
        return str(n)
    else:
        return to_binary(n // 2) + str(n % 2)

# Проверка работы функции
number = int(input('Введите целое неотрицательное число: '))

print('Двоичная запись:', to_binary(number))
````

Функция to_binary() рекурсивно делит число на 2 и получает остаток от деления. Остатки формируют двоичную запись числа.

Базовый случай предусмотрен для чисел меньше 2, то есть для 0 и 1. Это необходимо, чтобы функция могла правильно обработать число 0. Если остановиться только на числе 1, при вводе нуля рекурсия не завершится корректно.

- Built-in — встроенные имена Python.

```
x = 'Global'def outer():    x = 'Enclosing'    def inner():        x = 'Local'        print(x)    inner()outer()
```

29. Какие правила порядка аргументов в сигнатуре функции нужно соблюдать? Приведите корректный пример и один заведомо ошибочный.

При объявлении функции сначала указываются обязательные позиционные параметры, затем параметры со значениями по умолчанию, далее *args, именованные параметры и **kwargs.

Корректный пример:
```
def example(a, b=10, *args, c=20, **kwargs):    print(a, b, args, c, kwargs)
```

Ошибочный пример:
```
def example(a=10, b):    print(a, b)
```


Ошибка возникает потому, что обязательный параметр b расположен после параметра a, у которого есть значение по умолчанию.

30. В каких случаях стоит явно требовать именованные аргументы (через * в сигнатуре)? Зачем это делают в библиотеках?

Именованные аргументы стоит требовать, если в функции много параметров или важно, чтобы пользователь явно указывал их назначение. Это делает код понятнее и снижает вероятность ошибок.

Пример:
```
def create_user(name, *, role='user', active=True):    print(name, role, active)create_user('Анна', role='admin', active=False)
```


Символ * запрещает передавать параметры role и active позиционно. В библиотеках это помогает сделать использование функций более понятным и безопасным.
