## Анализ задания, получение данных

Файл p7_data_06.csv расположен в репозитории GitHub. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Реализуем чтение всего файла построчно. Используем метод `open()`, а затем будем обходить файл как коллекцию с использованием цикла `for line in F`.

Для начала выведем первые несколько строк файла:

````python
F = open('p7_data_06.csv')

n_to_print = 10

for line in F:
    n_to_print -= 1
    if n_to_print > 0:
        print(line)

F.close()
````

Теперь разделим каждую строку на отдельные элементы с помощью метода `split(',')`. Первая строка содержит заголовки столбцов.

````python
F = open('p7_data_06.csv')

header = F.readline().strip().split(',')

print(header)
print('-' * 20)

n_to_print = 10

for line in F:
    data = line.strip().split(',')

    n_to_print -= 1
    if n_to_print > 0:
        print(data)

F.close()
````

Сформируем список `all_targets`, в котором каждая цель будет представлена в виде словаря.

````python
F = open('p7_data_06.csv')

header = F.readline().strip().split(',')

# Список всех целей
all_targets = []

for line in F:
    data = line.strip().split(',')

    target = {}

    for i, key in enumerate(header):
        target[key] = data[i]

    # Преобразуем значения к нужным типам
    target['Param1'] = int(target['Param1'])
    target['Param2'] = float(target['Param2'])
    target['Param3'] = float(target['Param3'])
    target['HasWeapon'] = target['HasWeapon'] == 'true'

    all_targets.append(target)

print(all_targets[:10])

F.close()
````

## Решение задачи 1

Первая задача: определить среднее значение Param1 по каждой из категорий.

Для решения задачи переберём все элементы списка all_targets и вычислим сумму Param1 и количество целей для каждой категории.



````python
# Словари для хранения суммы и количества целей
sum_param1 = {}
count_category = {}

for target in all_targets:
    category = target['Category']

    if category not in sum_param1:
        sum_param1[category] = 0
        count_category[category] = 0

    sum_param1[category] += target['Param1']
    count_category[category] += 1

# Вычислим среднее значение
for category in sum_param1:
    avg = sum_param1[category] / count_category[category]

    print(f'Категория: {category}')
    print(f'Среднее значение Param1: {avg:.2f}')
````

## Решение задачи 2

Вторая задача: определить сумму параметров Param2 для каждой категории.

Переберём все цели и сложим значения Param2 отдельно для каждой категории.

````python
# Словарь для хранения сумм Param2
sum_param2 = {}

for target in all_targets:
    category = target['Category']

    if category not in sum_param2:
        sum_param2[category] = 0

    sum_param2[category] += target['Param2']

for category in sum_param2:
    print(f'Категория: {category}')
    print(f'Сумма Param2: {sum_param2[category]:.2f}')
````

## Решение задачи 3

Третья задача: определить, сколько целей снабжены оружием, а сколько — нет.

Для решения задачи будем использовать условный оператор if и два счётчика.

````python
# Счётчики целей
weapon_yes = 0
weapon_no = 0

for target in all_targets:
    if target['HasWeapon']:
        weapon_yes += 1
    else:
        weapon_no += 1

print('Количество целей с оружием:', weapon_yes)
print('Количество целей без оружия:', weapon_no)
````

## Решение задачи 4

Четвёртая задача: найти объект с максимальным значением Param1.

Переберём все элементы списка all_targets и выберем объект с наибольшим значением параметра Param1.

````python
# Переменная для хранения объекта
target_max = all_targets[0]

for target in all_targets:
    if target['Param1'] > target_max['Param1']:
        target_max = target

print('Объект с максимальным значением Param1:')
print(target_max)
print('Максимальное значение Param1:', target_max['Param1'])
````
