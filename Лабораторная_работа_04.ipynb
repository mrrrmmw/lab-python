В файле data_v2.txt содержится последовательность символов, состоящая из букв латинского алфавита и десятичных цифр.

Необходимо найти минимальное число, кратное 3. Под числом понимается непрерывная последовательность цифр.

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
````

После загрузки файла выполним чтение всего файла в строку:

````python
f = open('data_v2.txt')
s = f.read().strip()

print(len(s))
````

Метод

```
strip()
```
используется для удаления лишних символов переноса строки и пробелов в начале и конце файла.

## Обработка символьной последовательности

Будем последовательно обходить строку. Если текущий символ является цифрой, будем добавлять его к строке `target`.

Когда последовательность цифр заканчивается, преобразуем полученную строку в число с помощью `int()` и проверим:

*   является ли число кратным 3
*   является ли оно минимальным среди найденных чисел

Для проверки кратности 3 используется остаток от деления:

Для проверки кратности 3 используется остаток от деления:


```
x % 3 == 0
```



````python
target = ''
minx = None

for c in s:
    if c.isdigit():
        target += c
    else:
        if target != '':
            x = int(target)

            if x % 3 == 0:
                if minx is None or x < minx:
                    minx = x

            target = ''

if target != '':
    x = int(target)

    if x % 3 == 0:
        if minx is None or x < minx:
            minx = x

print('Минимальное число, кратное 3 =', minx)
````

1. Опишите назначение, сходства и различия функций чтения из
текстового файла: `read( )`, `readline( )`, `readlines( )`.

Все три метода используются для чтения данных из текстового файла, но отличаются объёмом считываемой информации:

* `read()` — считывает всё содержимое файла или указанное количество символов и возвращает строку.
* `readline()` — считывает одну строку файла.
* `readlines()` — считывает все строки файла и возвращает их в виде списка строк.

2. Как прочитать многострочный текстовый файл не используя явно
методы `read( )`, `readline( )`, `readlines( )`?

Можно использовать цикл for, который последовательно перебирает строки файла:


```
with open('data.txt') as f:
    for line in f:
        print(line)
```
В этом случае строки считываются последовательно без явного вызова методов `read()`, `readline()` или `readlines()`.

3. Какова задача закрытия файла?

Закрытие файла освобождает ресурсы операционной системы, связанные с открытым файлом, и завершает работу программы с этим файлом.

4. Что такое позиция считывания файла? Где первоначально находится
позиция считывания при открытии файла ввода?

Позиция считывания — это место в файле, с которого будет происходить следующее чтение данных.

При открытии файла для чтения первоначальная позиция находится в начале файла, то есть перед первым символом.

5. В каком режиме открывается файл, если требуется записать в него
данные, но при этом требуется оставить существующее содержимое файла
нетронутым? В какую часть файла данные записываются при этом?

Для этого используется режим a (append — добавление):


```

```
Существующее содержимое файла сохраняется, а новые данные добавляются в конец файла.
